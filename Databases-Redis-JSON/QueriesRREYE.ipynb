```python
import json

# Load the JSON database from a file
def load_database(filename="json_database.json"):
    try:
        with open(filename, "r") as file:
            return json.load(file)
    except FileNotFoundError:
        return {"users": []}

# Save the JSON database to a file
def save_database(database, filename="json_database.json"):
    with open(filename, "w") as file:
        json.dump(database, file, indent=4)

# CREATE: Add a new user
def add_user(user_id, name, email):
    database = load_database()
    database["users"].append({
        "user_id": user_id,
        "name": name,
        "email": email,
        "orders": []
    })
    save_database(database)

# READ: Get user details by user_id
def get_user(user_id):
    database = load_database()
    for user in database["users"]:
        if user["user_id"] == user_id:
            return user
    return None

def get_users_by_name(name):
    database = load_database()
    matched_users = [user for user in database["users"] if user["name"] == name]
    return matched_users

def get_users_with_any_item(*items):
    database = load_database()
    items_lower = [item.lower() for item in items]
    matched_users = []

    for user in database["users"]:
        products = [order["product"].lower() for order in user.get("orders", [])]
        if any(item in products for item in items_lower):
            matched_users.append(user)

    return matched_users

def get_users_with_all_items(*items):
    database = load_database()
    items_lower = [item.lower() for item in items]
    matched_users = []

    for user in database["users"]:
        products = [order["product"].lower() for order in user.get("orders", [])]


        if all(item in products for item in items_lower):
            matched_users.append(user)

    return matched_users

# UPDATE: Update user information
def update_user(user_id, name=None, email=None):
    database = load_database()
    for user in database["users"]:
        if user["user_id"] == user_id:
            if name:
                user["name"] = name
            if email:
                user["email"] = email
            save_database(database)
            return user
    return None

# DELETE: Remove a user by user_id
def delete_user(user_id):
    database = load_database()
    database["users"] = [user for user in database["users"] if user["user_id"] != user_id]
    save_database(database)

# CREATE: Add an order for a user
def add_order(user_id, order_id, product, amount):
    database = load_database()
    for user in database["users"]:
        if user["user_id"] == user_id:
            user["orders"].append({
                "order_id": order_id,
                "product": product,
                "amount": amount
            })
            save_database(database)
            return user["orders"]
    return None

# READ: Get orders for a user
def get_orders(user_id):
    database = load_database()
    for user in database["users"]:
        if user["user_id"] == user_id:
            return user["orders"]
    return None

# UPDATE: Update an order
def update_order(user_id, order_id, product=None, amount=None):
    database = load_database()
    for user in database["users"]:
        if user["user_id"] == user_id:
            for order in user["orders"]:
                if order["order_id"] == order_id:
                    if product:
                        order["product"] = product
                    if amount:
                        order["amount"] = amount
                    save_database(database)
                    return order
    return None

# DELETE: Remove an order
def delete_order(user_id, order_id):
    database = load_database()
    for user in database["users"]:
        if user["user_id"] == user_id:
            user["orders"] = [order for order in user["orders"] if order["order_id"] != order_id]
            save_database(database)
            return user["orders"]
    return None
```
