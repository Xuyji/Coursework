#### Problem statement.

Given a list of possible passowrds (`possible_passwords`), identify which of these passwords are valid passwords and which of these passwords are not valid passwords. Valid passwords should satisfy the followin criteria:

1. must start with a letter;
2. must be between 8 and 13 characters long;
3. contain:
  * at least three letters (small or capital);
  * at least one digit;
  * at least one of the following special **seven** special charaters: (@ # $ % _ & *)
  * do not contain any characters other than letters, digits, and the characters from the list of special characters.

The target output is provided.

```python
import re

def possible_passwords_validation(possible_passwords):
    valid_passwords = []
    invalid_passwords = []
    special_characters = "@#$%_&*"

    for password in possible_passwords:
        if len(password) < 8 or len(password) > 13:
            invalid_passwords.append(password)
            continue
        elif not password[0].isalpha():
            invalid_passwords.append(password)
            continue

        letter_count = sum(c.isalpha() for c in password)
        digit_count = sum(c.isdigit() for c in password)
        special_char_count = sum(c in special_characters for c in password)

        if letter_count < 3 or digit_count < 1 or special_char_count < 1:
            invalid_passwords.append(password)
            continue

        if any(not c.isalnum() and c not in special_characters for c in password):
            invalid_passwords.append(password)
            continue


        valid_passwords.append(password)

    return valid_passwords, invalid_passwords
```

```python
#now the output
valid_passwords, invalid_passwords = possible_passwords_validation(possible_passwords)

print("Valid passwords:")
print(valid_passwords)
print("Invalid passwords:")
print(invalid_passwords)
```
