## Selected Capital Cities and Hemispheres

1. **Washington, D.C. (USA)**
   - Hemisphere: Northern, Western
   - Latitude: 38.9072
   - Longitude: -77.0369

2. **Brasília (Brazil)**
   - Hemisphere: Southern, Western
   - Latitude: -15.7939
   - Longitude: -47.8828

3. **Canberra (Australia)**
   - Hemisphere: Southern, Eastern
   - Latitude: -35.2809
   - Longitude: 149.1300

4. **Tokyo (Japan)**
   - Hemisphere: Northern, Eastern
   - Latitude: 35.6762
   - Longitude: 139.6503


```python
import requests

API_KEY = "133490cb972073e36b9963d474a64019"
```

```python
cities = {
    "Washington, D.C.": (38.9072, -77.0369),
    "Brasília": (-15.7939, -47.8828),
    "Canberra": (-35.2809, 149.1300),
    "Tokyo": (35.6762, 139.6503)
}
```

```python
weather_data = {}

for city, (lat, lon) in cities.items():
    url = f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={API_KEY}"
    response = requests.get(url)
    data = response.json()
    temperature = data["main"]["temp"]
    weather_data[city] = temperature
    print(f"{city}: {temperature}°C")
```

```python
coldest_city = min(weather_data, key=weather_data.get)
coldest_temp = weather_data[coldest_city]

print(f"The capital city with the lowest temperature is {coldest_city} at {coldest_temp}°C.")
```
