### Distribution of the ride duration

* Convert seconds into minutes
* Binning

```python
data["started_at"] = pd.to_datetime(data["started_at"])
```

```python
data["ended_at"] = pd.to_datetime(data["ended_at"])
```

```python

data["tripduration"] = (data['ended_at'] - data['started_at']).dt.total_seconds() / 60
data["tripduration"]
```

```python
(data.tripduration // 60).value_counts().sort_values()
```

```python
data['tripminutes'] = data['tripduration'] // 60
```

```python
sns.displot(data=data,x="tripminutes", bins=range(1,61), aspect=10/5);
```

### Distribution of the ride start hour (to confirm that CitiBike is used for work commute)

* Get the start hour


```python
data.dtypes
```

```python
data['starthour'] = data['started_at'].dt.hour.astype('category')
```

```python
data['endhour'] = data['ended_at'].dt.hour.astype('category')
```

```python
data['startday'] = data['started_at'].dt.day.astype('category')
data['weekday'] = data['started_at'].dt.weekday.astype('category')
```

```python
data.dtypes
```

```python
plt.figure(figsize=(12,5))
sns.countplot(data=data, x="starthour")
```

```python
plt.figure(figsize=(12,5))
sns.countplot(data=data, x="startday")
```
