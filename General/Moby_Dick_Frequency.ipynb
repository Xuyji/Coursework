```python
import os
import string
from collections import Counter
import pandas as pd
```

```python
moby_dick_folder = "/content/moby_dick_chapters"


moby_dick_files = [os.path.join(moby_dick_folder, f"moby-{str(i).zfill(3)}.txt")
                   for i in range(1, 135) if os.path.exists(os.path.join(moby_dick_folder, f"moby-{str(i).zfill(3)}.txt"))]


book_text = ""
for file in moby_dick_files:
    with open(file, "r", encoding="utf-8") as f:
        book_text += f.read().lower() + " "


book_text = book_text.translate(str.maketrans("", "", string.punctuation))


words = book_text.replace("\n", " ").split()

bip39_file_path = "/content/bip-0039.txt"
with open(bip39_file_path, "r", encoding="utf-8") as f:
    bip39_words = set(f.read().splitlines())


bip39_counter = Counter(word for word in words if word in bip39_words)

sorted_bip39_words = bip39_counter.most_common()

bip39_df = pd.DataFrame(sorted_bip39_words, columns=["Word", "Frequency"])

pd.set_option("display.max_rows", None)
pd.set_option("display.max_colwidth", None)

print(bip39_df)
bip39_df.to_csv("bip39_word_frequencies.csv", index=False)

from google.colab import files
files.download("bip39_word_frequencies.csv")

```
