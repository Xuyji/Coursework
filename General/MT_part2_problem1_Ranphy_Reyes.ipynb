```python

# Question 1
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/pycaret/pycaret/master/datasets/diamond.csv')
df.head()
```

```python
#Question 2
df.groupby(['Cut', 'Color'])['Carat Weight'].mean()
```

```python
#Question 3
df.groupby(['Clarity', 'Color']).size()

```

```python
#Question 4 (What is the average price depending on the cut and color?)
pivot_table = df.pivot_table(index='Cut', columns='Color', values='Price', aggfunc='mean')
pivot_table
```


 The price is all about balancing the diamond's 4Cs . If you want like the "Ideal" cuts and "D" color grades, you'll pay a premium for that perfection. But if you're willing to compromise a bit on the brightness or accept a slight hint of color with a "Fair" cut or an "I" color grade, you can find options that are significantly more budget friendly.
