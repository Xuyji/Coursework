# Question 1

```python
import math
import pandas as pd
import numpy as np
```

```python
N = 1000000

tf_data ={
    'Document X':{'apple': 50, 'banana': 30, 'cherry': 10, 'date': 20, 'elderberry': 5, 'fig': 15, 'grape': 25, 'honeydew': 10 },
    'Document Y':{'apple': 40, 'banana': 35, 'cherry': 5,  'date': 15, 'elderberry': 10,'fig': 20, 'grape': 30, 'honeydew': 8 },
    'Document Z':{'apple': 60, 'banana': 20, 'cherry': 25, 'date': 25, 'elderberry': 8, 'fig': 10, 'grape': 40, 'honeydew': 12 }
}

df_data = {
    'apple': 20000,
    'banana': 15000,
    'cherry': 5000,
    'date': 10000,
    'elderberry': 2000,
    'fig': 7500,
    'grape': 25000,
    'honeydew': 9000
}
```

```python
idf_data = {term: math.log10(N / df) for term, df in df_data.items()}

for doc, tfs in tf_data.items():
  print(f"{doc}:")
  for term, tf in tfs.items():
    df = df_data[term]
    idf = idf_data[term]
    tf_idf = tf * idf
    print(f"  {term}: TF={tf}, IDF={idf}, TF-IDF={tf_idf}")
  print()
```

# Question 2

```python
vocabulary = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H']
documents = {
    'Doc1': 'AAABDF',
    'Doc2': 'AABEG',
    'Doc3': 'CEH',
    'Doc4': 'AADG',
    'Doc5': 'BBCEEHH'
}

num_docs = len(documents)

tf_matrix = np.zeros((len(vocabulary), num_docs))

for j, (doc_name, doc_text) in enumerate(documents.items()):
    for i, term in enumerate(vocabulary):
        count = doc_text.count(term)

        tf_matrix[i, j] = (1 + math.log10(count)) if count > 0 else 0

df_array = np.sum(tf_matrix > 0, axis=1)
idf_array = np.array([math.log10(num_docs / df) if df > 0 else 0 for df in df_array])


td_matrix = tf_matrix * idf_array[:, np.newaxis]

td_df = pd.DataFrame(td_matrix, index=vocabulary, columns=documents.keys())
print("Term-Document Matrix (Log-Normalized TF * IDF):")
print(td_df.round(4))
print("\n")


tt_matrix = np.dot(td_matrix, td_matrix.T)
tt_df = pd.DataFrame(tt_matrix, index=vocabulary, columns=vocabulary)
print("Term-Term Correlation Matrix:")
print(tt_df.round(4))
```

# Question 3
When figuring out Term Frequency and Inverse Document Frequency, there isn't just one right way to do the math. For TF, you can use raw frequency counts, or even a simple Boolean method that just checks if a word is present 1 or absent 0. But since longer documents naturally repeat words more, we usually apply log-normalization or divide the count by the document's maximum TF to level the playing field.   For IDF, the standard logarithmic equation works well, but it can cause divide-by-zero errors if a term is totally missing. To fix this, we use smoothed IDF, or we apply probabilistic IDF to weigh the ratio of documents that have the term against those that don't. By mixing and matching these different approaches, search systems build various TF-IDF weighting schemes, which are often categorized using frameworks like SMART notation

# Question 4
The Boolean model is pretty rigid because it relies completely on exact set theory logic. A document either perfectly matches your AND/OR/NOT conditions, or it doesn't show up at all. Term weighting models, like the Vector Space Model, take a much more nuanced approach by treating everything as vectors and calculating a continuous similarity score.   The huge advantage of term weighting is that it gives you a ranked list of results instead of an unsorted data dump. It naturally prioritizes rare, meaningful words and allows for partial matches, so if a document is missing one search term but heavily features the rest, you still get to see it. The strict binary logic of the Boolean model completely fails at this.   On the flip side, term weighting requires a lot more processing power to handle the complex matrix math. It also takes away the strict control that power users like about Boolean searches you can't easily force a term-weighting system to definitively include or exclude a specific word.
