```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import datetime
```

```python

curl = 'https://raw.githubusercontent.com/jacobstallone/Time_Series_ARIMA--Blog-and-code-/master/AirPassengers.csv'
df = pd.read_csv(curl)


```

```python
print(df.info())
df['Month'] = pd.to_datetime(df['Month'])
df.set_index('Month', inplace=True)
df.rename(columns={'#Passengers': 'Passengers'}, inplace=True)
print(df.info())
df.head()
```

 Key transformations involved converting the 'Month' column to a DatetimeIndex and ensuring the 'Passengers' column contains sequential numerical data recorded over time . the data successfully cleaned and structured, it is now ready for the application of standard time series analysis techniques like forecasting, trend analysis, and seasonality decomposition . it


```python
#Plot the time series number of air passengers over time
df.plot(figsize=(12,8))
plt.show()
```

```python
#Question 2
rolling_mean_12 = df.rolling(window=12).mean()
rolling_mean_24 = df.rolling(window=24).mean()
plt.figure(figsize=(12,8))
plt.plot(df.index, df['Passengers'], label='Original Data')
plt.plot(rolling_mean_12.index, rolling_mean_12['Passengers'], label='Rolling Mean (12 months)')
plt.plot(rolling_mean_24.index, rolling_mean_24['Passengers'], label='Rolling Mean (24 months)')
plt.xlabel('Date')
plt.ylabel('Number of Passengers')
plt.title('Original Data and Rolling Means')
plt
plt.legend()
plt.show()
```
