#### Exercise 1

Write a program to create a Pivot table and find the total sale amount Region and manager wise.


```python
pd.pivot_table(df,index = ["Region","Manager"], values = ["Sale_amt"],aggfunc=np.sum)
```

#### Exercise 2

Write a program to create a Pivot table and find the total sale amount Region, manager, and sales man wise.

```python
pd.pivot_table(df,index=["Region","Manager","SalesMan"], values="Sale_amt", aggfunc=np.sum)
```

#### Exercise 3

Write a program to create a Pivot table and find the item wise unit sold.

```python
pd.pivot_table(df,index=["Item"], values="Units", aggfunc=np.sum)
```

#### Exercise 4

Write a program to create a Pivot table and count the manager wise sale and mean value of sale amount.

```python
pd.pivot_table(df,index=["Manager"],values=["Sale_amt"],aggfunc=[np.mean,len])
```

#### Exercise 5

Write a program to create a Pivot table and find the total sale amount region wise, manager wise, sales man wise where Manager = "Douglas".

```python
table = pd.pivot_table(df,index=["Region","Manager","SalesMan"], values="Sale_amt")
print(table.query('Manager == ["Douglas"]'))
```

#### Exercise 6

Write a program to create a Pivot table and find the maximum and minimum sale value of the items.



```python
table = pd.pivot_table(df, index='Item', values='Sale_amt', aggfunc=[np.max, np.min])
table
```

#### Exercise 7

Write a program to create a 3-year time series using three months frequency. Start date is Janury 1, 2021



```python
time_series = pd.date_range('1/1/2021', periods = 36, freq='3M')
print("Time series using three months frequency:")
print(time_series)
```

#### Exercise 8
Write a program to check if a day is a business day (weekday) or not.



```python
def is_business_day(date):
    return bool(len(pd.bdate_range(date, date)))
```

```python
print("Check busines day or not?")
print('2020-12-01: ',is_business_day('2020-12-01'))
print('2020-12-06: ',is_business_day('2020-12-06'))
print('2020-12-07: ',is_business_day('2020-12-07'))
print('2020-12-08: ',is_business_day('2020-12-08'))
print('2024-10-21: ',is_business_day('2024-10-21'))
print('2024-10-20: ',is_business_day('2024-10-20'))
```
