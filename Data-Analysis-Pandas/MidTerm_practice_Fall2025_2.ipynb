### Step 1: import the libraries necessary for the assignment

```python
%matplotlib inline
import requests
import pandas as pd
import numpy as np
```

### Step 2: import the NYC schools dataset on 2017-2018 Class Size Report Borough K-8 Average Class Size
https://data.cityofnewyork.us/Education/2017-2018-Class-Size-Report-Borough-K-8-Average-Cl/3w89-mz4u/about_data


---



```python
# Data set:
url = 'https://data.cityofnewyork.us/resource/3w89-mz4u.json'
results = requests.get(url).json()
```

```python
df = pd.DataFrame(results)
df
```

```python
df.dtypes
```

### Step 3: convert `object` data.
* `borough`, `grade_level`, `program_type` attribute to categorical data type
* `number_of_students` and `number_of_classes` to integer
* `average_class_size` to float



```python
df['borough'] = df['borough'].astype('category')
df['grade_level'] = df['grade_level'].astype('category')
df['program_type'] = df['program_type'].astype('category')
```

```python
df['number_of_students'] = df['number_of_students'].astype('int')
df['number_of_classes'] = df['number_of_classes'].astype('int')
```

```python
df['average_class_size'] = df['average_class_size'].astype('float')
```

```python
df.dtypes
```

### Step 4: Pivot table use
Compute the number of students per program in each borough.

```python
pivot = pd.pivot_table(df,
                       values='number_of_students',
                       index=['borough'], # rows
                       columns=['program_type'], # columns
                       aggfunc=np.sum) # aggregation function
pivot
```

```python
flattened = pd.DataFrame(pivot.to_records())
flattened
```

### Step 5: Pivot table use (addional attributes)
Compute the ratio of the students in ICT and G&T program to GenEd students for each borough

```python
flattened["ICTtoGenED"] = flattened["ICT"] / flattened["Gen Ed"]
flattened["G&TtoGenED"] = flattened["G&T"] / flattened["Gen Ed"]

flattened
```

### Step 6: Pivot table use (sorting)
Sort the boroughs by
* the number of students in the GenEd program;
* the number of students in the ICT program;
* the ratio of the number of students in the ICT program over the number of students in the GenEd program.

**What comclusions can be made from these three sorted lists?**

```python
GenEd = flattened[["borough","Gen Ed"]]
GenEd.sort_values(by="Gen Ed", ascending=False)
```

```python
ICT = flattened[["borough","ICT"]]
ICT.sort_values(by="ICT", ascending=False)
```

```python
RatioICT = flattened[["borough","ICTtoGenED"]]
RatioICT.sort_values(by="ICTtoGenED",ascending=False)
```
